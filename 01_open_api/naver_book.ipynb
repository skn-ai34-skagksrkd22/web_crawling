```python
import urllib.parse
import urllib.request
import json
import mysql.connector
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
client_id = os.getenv('naver_client_id')
client_secret = os.getenv('naver_secret')
```

```python
searchText = urllib.parse.quote('SK')
```

```python
url = "https://openapi.naver.com/v1/search/book.json?query="+searchText+"&display=100"

request = urllib.request.Request(url)
request.add_header('X-Naver-Client-Id', client_id)
request.add_header('X-Naver-Client-Secret', client_secret)
```

```python
response = urllib.request.urlopen(request)
response_body = response.read()
response_body = json.loads(response_body)
book_list = response_body['items']
```

```python
book_list
```

```python
with mysql.connector.connect( 
    host = 'localhost',
    user = 'root',
    password = '2401',
    database = 'bookdb'
) as connection:
    with connection.cursor() as cursor:
        sql = "insert into naver_book" \
            "(book_title, book_image, author, publisher, isbn, book_description, pub_date)" \
            "values (%s, %s, %s, %s, %s, %s, %s)"
        
        for book_info in book_list:
            values = (
                book_info['title'],
                book_info['image'],
                book_info['author'],
                book_info['publisher'],
                book_info['isbn'],
                book_info['description'],
                book_info['pubdate']
            )
            
            cursor.execute(sql, values)
    connection.commit()
```
